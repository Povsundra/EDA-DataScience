- Check kind of missing values (MCAR, MAR,or MNAR)


````python
import plotly.graph_objects as go

# Original data
df_income_all = df.copy()

# Data after dropping missing Income
df_income_dropped = df.dropna(subset=['Income'])
````

````python
fig_income = go.Figure()

# Before dropping NA
fig_income.add_trace(
    go.Histogram(
        x=df_income_all['Income'],
        name="Before dropping NA",
        showlegend=True
    )
)

# After dropping NA
fig_income.add_trace(
    go.Histogram(
        x=df_income_dropped['Income'],
        name="After dropping NA",
        showlegend=True,
        visible="legendonly"   # cleaner interaction
    )
)

fig_income.update_layout(
    barmode='overlay',
    title="Income Distribution Before and After Removing Missing Values",
    xaxis=dict(title="Income"),
    yaxis=dict(title="Count"),              
    width=600,
    height=450
)

fig_income.update_traces(opacity=0.6)

fig_income.show()
````

Basedo on overlay histograms it seems that the distribution very similar, the missing values in the Income column are likely MCAR (Missing Completely at Random). 

Therefore, removing records with missing income values is unlikely to introduce bias into the analysis. 

- TEsting: Compare Distributions of OTHER VARIABLES

````python
df['Income_missing'] = df['Income'].isna()
````

````python
fig = go.Figure()

fig.add_trace(go.Histogram(
    x=df.loc[df['Income_missing'] == False, 'total_purchases'],
    name="Income Observed"
))

fig.add_trace(go.Histogram(
    x=df.loc[df['Income_missing'] == True, 'total_purchases'],
    name="Income Missing",
    visible="legendonly"
))

fig.update_layout(
    barmode='overlay',
    title="Total Purchases Distribution by Income Missingness",
    xaxis_title="Total Purchases",
    yaxis_title="Count"
)

fig.update_traces(opacity=0.6)
fig.show()
````

````python
# statistical test 
from scipy.stats import ks_2samp

vars_to_check = ['expenses', 'total_purchases', 'kids']

for v in vars_to_check:
    x1 = df.loc[df['Income_missing'] == False, v].dropna()
    x2 = df.loc[df['Income_missing'] == True, v].dropna()
    
    stat, p = ks_2samp(x1, x2)
    print(f"{v}: KS p-value = {p:.4f}")

# evidence: simmilar distributions and p > 0.05 => MCAR safe to drop 
````
